Drop rows containing missing values.

Fill numeric missing values with mean.

```python
data = {
    "Name": ["Rahul", "Bhanu", "Satyam", "Amit", "Keshav", "Raman", "kunal"],
    "Age": [25, 22, np.nan, 32,36, np.nan, 25]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
mean = np.mean(df['Age'])
print(mean)
```

```python
df.fillna(mean)
```

Fill missing values with median.

```python
data = {
    "Name": ["Rahul", "Bhanu", "Satyam", "Amit", "Keshav", "Raman", "kunal"],
    "Age": [25, 22, np.nan, 32,36, np.nan, 25]
}

df = pd.DataFrame(data)
df
```

```python
series = pd.Series(df['Age'])
median = series.median()
median
```

```python
df.fillna(median)
```

Fill categorical missing values with mode.

```python
data = {
    "Name": ["Rahul", "Bhanu", "Satyam", "Amit", "Keshav", "Raman", "kunal"],
    "Age": [25, 22, np.nan, 32,36, np.nan, 25]
}

df = pd.DataFrame(data)
df
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mode()[0])
df
```

Forward-fill missing values.

```python
data = {
    "Name": ["Rahul", "Bhanu", "Satyam", "Amit", "Keshav", "Raman", "kunal"],
    "Age": [25, 22, np.nan, 32,36, np.nan, 25]
}

df = pd.DataFrame(data)
df
```

```python
df['Age'] = df['Age'].ffill()
df
```

Backward-fill missing values.

```python
data = {
    "Name": ["Rahul", "Bhanu", "Satyam", "Amit", "Keshav", "Raman", "kunal"],
    "Age": [25, 22, np.nan, 32,36, np.nan, 25]
}

df = pd.DataFrame(data)
df
```

```python
df['Age'] = df['Age'].backfill()
df
```
