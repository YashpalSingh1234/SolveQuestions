Perform matrix multiplication using @.

Calculate cosine similarity.

```python
vec1 = np.array([2,3,5])
vec2 = np.array([-2,-3,-5])

cosine_similarity = np.dot(vec1, vec2) / (np.linalg.norm(vec1) * np.linalg.norm(vec2))
print(cosine_similarity)
```
