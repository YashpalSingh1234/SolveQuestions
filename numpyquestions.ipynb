Normalize every column.

```python
'''
for normlizeation we use min-max normlization
formula:   (x-min)/(max-min)
'''

arr = np.array([
    [1, -2, 3, 4],
    [5, 6, 2, 8],
    [9, -10, 1, 12],
    [13, 14, 15, -16]
])

column_min = np.min(arr, 0)
column_max = np.max(arr, 0)
print(column_min, column_max)


norm_arr = (arr-column_min)/ (column_max-column_min)
print(norm_arr)
```

Standardize every column.

```python
"""
For standardize every column
transform data through Z-score standardization

Formula:  Z = (x − mean) / standard deviation
"""

arr = np.array([
    [1, -2, 3, 4],
    [5, 6, 2, 8],
    [9, -10, 1, 12],
    [13, 14, 15, -16]
])


column_mean = np.mean(arr, 0)
print(column_mean)
column_std = np.std(arr, 0)
print(column_std)

arr = (arr - column_mean)/ column_std
print(arr)
```

Implement simple linear regression using NumPy.

```python
x= np.array([1,2,3,4,5])
y= np.array([2,4,5,8,10])

# Mean

x_mean = np.mean(x)
y_mean = np.mean(y)


# Slope (b1)

b1 = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)

# Intercept (b0)
b0 = y_mean - b1 * x_mean

# Predictions
y_pred = b0 + b1 * x

print("slope: ", b1)
print("Intercept: ", b0)
print("Predictions: ", y_pred)
```
