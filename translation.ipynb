एक छोटे से शहर में आरव नाम का एक लड़का रहता था। वह एक साधारण परिवार से था, लेकिन उसके सपने बिल्कुल साधारण नहीं थे। बचपन से ही उसे नई चीजें सीखने और अलग-अलग समस्याओं को हल करने में बहुत रुचि थी। जब वह स्कूल में था, तब उसके दोस्तों को खेलना और टीवी देखना पसंद था, लेकिन आरव अक्सर अपना खाली समय कंप्यूटर पर बिताता था। वह इंटरनेट पर अलग-अलग चीजों के बारे में पढ़ता रहता था और कभी-कभी ऐसी चीजें सीखने की कोशिश करता था जिन्हें उसके स्कूल में अभी तक पढ़ाया भी नहीं गया था।

आरव के पिता एक छोटी सी दुकान चलाते थे और उसकी माँ घर संभालती थीं। परिवार की आर्थिक स्थिति बहुत अच्छी नहीं थी, इसलिए आरव जानता था कि उसे अपने भविष्य के लिए खुद मेहनत करनी होगी। उसके माता-पिता हमेशा उससे कहते थे कि अगर वह ईमानदारी से मेहनत करेगा और अपने लक्ष्य पर ध्यान देगा, तो एक दिन वह अपने जीवन में कुछ अच्छा जरूर कर पाएगा।

स्कूल खत्म होने के बाद आरव ने कंप्यूटर साइंस पढ़ने का फैसला किया। कॉलेज में आने के बाद उसे पहली बार महसूस हुआ कि केवल किताबों से पढ़ना काफी नहीं है। उसके आसपास बहुत से ऐसे विद्यार्थी थे जो उससे ज्यादा तेज थे और पहले से ही कई programming languages जानते थे। शुरुआत में आरव को लगा कि शायद वह बाकी छात्रों जितना अच्छा नहीं है। जब भी वह किसी programming problem को हल करने की कोशिश करता, उसे कई बार error मिलती थी। कभी उसका code काम नहीं करता था और कभी उसे समझ ही नहीं आता था कि गलती कहाँ हुई है।

एक दिन उसने अपने दोस्त रोहन से कहा, "मुझे लगता है कि मैं इस field के लिए बना ही नहीं हूँ। मैं कितनी भी कोशिश कर लूँ, मुझे चीजें समझ में नहीं आतीं।"

रोहन ने उसकी बात ध्यान से सुनी और कहा, "तुम्हारी सबसे बड़ी समस्या यह नहीं है कि तुम चीजें नहीं समझ सकते। तुम्हारी समस्या यह है कि तुम दूसरों से अपनी तुलना करते हो। अगर तुम हर दिन थोड़ा-थोड़ा सीखते रहोगे, तो कुछ महीनों बाद तुम्हें खुद फर्क दिखाई देगा।"

आरव को अपने दोस्त की बात सही लगी। उसने उसी दिन फैसला किया कि अब वह अपनी तुलना किसी और से नहीं करेगा। वह केवल अपने पिछले version से बेहतर बनने की कोशिश करेगा।

अगले दिन से उसने एक routine बनाया। वह सुबह जल्दी उठता, कॉलेज जाने से पहले एक घंटे programming की practice करता और कॉलेज से वापस आने के बाद दो घंटे अपने projects पर काम करता। रात में सोने से पहले वह उस दिन सीखी हुई चीजों को दोबारा revise करता। शुरुआत में यह routine उसके लिए बहुत मुश्किल था। कई बार वह थक जाता और सोचता कि आज practice छोड़ देता हूँ। लेकिन हर बार उसे अपने लक्ष्य की याद आती और वह कम से कम आधे घंटे के लिए ही सही, computer खोलकर बैठ जाता।

कुछ महीनों बाद उसके अंदर काफी बदलाव आने लगा। पहले जिन programming problems को देखकर वह डर जाता था, अब वह उन्हें छोटे-छोटे हिस्सों में बाँटकर solve करने लगा था। उसे समझ आने लगा था कि programming में हर problem का answer तुरंत मिलना जरूरी नहीं है। कई बार हमें एक ही problem को अलग-अलग तरीकों से solve करने की कोशिश करनी पड़ती है।

कॉलेज के अंतिम वर्ष में आरव ने एक ऐसा project बनाने का फैसला किया जो वास्तविक जीवन की किसी समस्या को हल कर सके। उसने अपने दोस्तों के साथ मिलकर एक ऐसा system बनाने की योजना बनाई जो students को उनके resumes को बेहतर बनाने में मदद कर सके। शुरुआत में project का idea बहुत simple था, लेकिन जैसे-जैसे उन्होंने काम करना शुरू किया, उन्हें पता चला कि project उनकी सोच से कहीं ज्यादा मुश्किल है।

उन्हें database की समस्या आई, फिर API में errors आने लगे। कभी frontend काम नहीं करता था और कभी backend response देने में बहुत ज्यादा समय लेता था। कई बार उनका पूरा system अचानक बंद हो जाता था। एक रात उन्होंने लगभग छह घंटे लगातार काम किया, लेकिन फिर भी उनका project सही तरीके से नहीं चल पाया।

उस रात आरव बहुत निराश हुआ। उसने laptop बंद किया और सोने चला गया। लेकिन अगले दिन सुबह उठने के बाद उसने project छोड़ने के बजाय यह पता लगाने का फैसला किया कि असली समस्या क्या थी। उसने अपने code को एक-एक करके check किया। कई घंटों की debugging के बाद उसे पता चला कि problem एक छोटी सी configuration mistake के कारण हो रही थी।

उसने उस mistake को ठीक किया और system फिर से चल पड़ा।

उस दिन आरव को एक बहुत महत्वपूर्ण बात समझ आई। कई बार हमें लगता है कि हमारी पूरी मेहनत बेकार जा रही है, लेकिन वास्तव में हम उस समय किसी ऐसी चीज के बारे में सीख रहे होते हैं जो भविष्य में हमारे बहुत काम आने वाली होती है।

कुछ समय बाद आरव ने अपना project online deploy कर दिया। उसने अपना portfolio बनाया और अलग-अलग companies में job के लिए apply करना शुरू किया। शुरुआत में उसे बहुत सारे rejection मिले। कुछ companies ने उसका resume देखकर जवाब ही नहीं दिया। कुछ interviews में वह technical questions का जवाब नहीं दे पाया। एक interview में तो interviewer ने उससे ऐसा सवाल पूछा जिसका answer उसे बिल्कुल नहीं पता था।

Interview से बाहर निकलने के बाद आरव को बहुत बुरा लगा। उसे लगा कि शायद उसने जितनी मेहनत की है, वह पर्याप्त नहीं है। वह घर वापस आया और अपने कमरे में बैठकर काफी देर तक सोचता रहा।

उसकी माँ ने जब उसे परेशान देखा तो पूछा, "क्या हुआ? आज तुम इतने चुप क्यों हो?"

आरव ने कहा, "मुझे लगता है कि मैं job के लिए अच्छा नहीं हूँ। मैं interview में कई सवालों के जवाब नहीं दे पाया।"

उसकी माँ मुस्कुराईं और बोलीं, "अगर आज तुम्हें सब कुछ आता होता, तो तुम्हें सीखने की जरूरत ही नहीं पड़ती। हारने और सीखने में फर्क समझो। अगर तुम हर failure के बाद रुक जाओगे, तो तुम कभी आगे नहीं बढ़ पाओगे। लेकिन अगर तुम अपनी गलती से सीखोगे, तो हर failure तुम्हें पहले से ज्यादा मजबूत बनाएगा।"

माँ की बात सुनकर आरव को थोड़ी हिम्मत मिली। उसने उस interview में पूछे गए सभी questions लिख लिए। अगले दिन से उसने उन topics को पढ़ना शुरू कर दिया जिन्हें वह नहीं जानता था। उसने अपने पुराने mistakes को भी दोबारा देखा और समझने की कोशिश की कि उसने कहाँ गलती की थी।

कुछ हफ्तों बाद उसे एक और interview का मौका मिला। इस बार interviewer ने उससे programming, machine learning और project-related questions पूछे। कुछ questions difficult थे, लेकिन आरव घबराया नहीं। जहाँ उसे answer पता था, उसने confidently explain किया और जहाँ उसे पूरा answer नहीं पता था, वहाँ उसने ईमानदारी से कहा कि उसे उस topic की पूरी जानकारी नहीं है, लेकिन वह उसे सीख सकता है।

Interview के अंत में interviewer ने उससे पूछा, "अगर आपको कोई ऐसी technology सीखनी हो जिसके बारे में आपको अभी कुछ भी नहीं पता, तो आप उसे कैसे सीखेंगे?"

आरव ने कुछ seconds सोचा और कहा, "सबसे पहले मैं उस technology की basics समझूँगा। उसके बाद छोटे examples बनाऊँगा। जब मुझे basic concepts समझ आ जाएंगे, तब मैं एक छोटा project बनाऊँगा। Project बनाते समय आने वाली problems को solve करते हुए मेरी understanding और मजबूत होगी।"

Interviewer ने उसकी बात सुनी और मुस्कुराया।

कुछ दिनों बाद आरव को उस company से job offer मिल गया।

जब उसने यह खबर अपने माता-पिता को बताई, तो उसकी माँ की आँखों में खुशी के आँसू आ गए। उसके पिता ने उसे गले लगाया और कहा, "हमें तुम पर हमेशा विश्वास था।"

आरव ने मुस्कुराते हुए कहा, "मुझे भी अब अपने ऊपर पहले से ज्यादा विश्वास है। लेकिन मैं अभी रुकना नहीं चाहता। यह सिर्फ शुरुआत है।"

नौकरी शुरू करने के बाद भी आरव ने सीखना बंद नहीं किया। वह जानता था कि technology तेजी से बदल रही है और अगर वह खुद को update नहीं रखेगा, तो कुछ सालों बाद वह पीछे रह सकता है। इसलिए उसने काम के साथ-साथ नई technologies सीखना जारी रखा। उसने अपने पुराने projects को improve किया, नए projects बनाए और अपनी communication skills पर भी काम करना शुरू किया।

कई साल बाद जब उसके पुराने कॉलेज के दोस्त उससे मिले, तो उन्होंने महसूस किया कि आरव में बहुत बदलाव आ चुका था। जो लड़का कभी दूसरों से अपनी तुलना करके परेशान हो जाता था, अब वही दूसरों को motivate कर रहा था।

एक दिन रोहन ने उससे पूछा, "तुम्हारे अंदर इतना बदलाव कैसे आया?"

आरव ने हँसते हुए कहा, "मैंने कोई जादू नहीं किया। मैंने बस एक चीज समझी—अगर हमें अपनी जिंदगी बदलनी है, तो हमें हर दिन छोटे-छोटे कदम उठाने पड़ते हैं। सफलता अचानक नहीं आती। वह उन छोटे decisions का result होती है जो हम हर दिन लेते हैं।"

रोहन ने कहा, "और अगर कोई व्यक्ति बार-बार fail हो?"

आरव ने जवाब दिया, "तो उसे failure को अपना दुश्मन नहीं, अपना teacher समझना चाहिए। जब तक हम अपनी गलतियों से सीखते रहते हैं, तब तक हम वास्तव में हार नहीं रहे होते।"

आरव की कहानी हमें यही सिखाती है कि हमारे रास्ते में कितनी भी मुश्किलें क्यों न आएँ, हमें अपने लक्ष्य की ओर बढ़ते रहना चाहिए। हमें दूसरों से अपनी तुलना करने के बजाय खुद को हर दिन थोड़ा बेहतर बनाने की कोशिश करनी चाहिए। कभी-कभी progress बहुत धीमी दिखाई देती है, लेकिन इसका मतलब यह नहीं है कि हम आगे नहीं बढ़ रहे हैं।

सबसे महत्वपूर्ण बात यह है कि हमें अपने ऊपर विश्वास बनाए रखना चाहिए। अगर हम लगातार सीखते रहें, अपनी गलतियों को स्वीकार करें और उन्हें सुधारने की कोशिश करें, तो एक दिन हम उस जगह पहुँच सकते हैं जहाँ हम खुद को देखना चाहते हैं।
