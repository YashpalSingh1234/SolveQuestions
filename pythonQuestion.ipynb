Count vowels in a string.

```python
vowels = ['a', 'e' ,'i','o','u', 'A','E','I','O','U']

def check_vowels(word:str):
    count = 0
    for c in word:
        if c in vowels:
            count += 1
        
    
    return count

print(check_vowels("Rahul"))
```

Count vowels and consonants separately.

```python
def vowels_consonent_count(word:str):
    vowels = 0
    consonent = 0

    for c in word:
        if c.isalpha():  
            if c in "aeiouAEIOU":
                vowels += 1
            else:
                consonent += 1
            
    return vowels, consonent


print(vowels_consonent_count("Rahul is a good boy"))
```

Find the frequency of each character.

```python
def frequency_of_charcter(string:str):
    data = {}
    for ch in string:
        if ch.isalpha():
            
            if ch not in data:
                data[ch] = 1
            else:
                data[ch] += 1
        
    return data

print(frequency_of_charcter("Rahul is a good boy"))
```

Find the first non-repeating character.

```python
def first_non_repeatingNumber(string:str):
    data = {}
    for ch in string:
        if ch.isalpha():
            if ch not in data:
                data[ch] = 1
            else:
                data[ch] += 1

    for key in data:
        if data[key] == 1:
            return key

print(first_non_repeatingNumber("Rahul is a good boy"))
```
