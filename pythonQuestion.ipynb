```python
def first_non_repeatingNumber(string:str):
    data = {}
    for ch in string:
        if ch.isalpha():
            if ch not in data:
                data[ch] = 1
            else:
                data[ch] += 1

    for key in data:
        if data[key] == 1:
            return key

print(first_non_repeatingNumber("Rahul is a good boy"))
```

Find all pairs whose sum equals a target.

```python
def func(ll:list, target:int):
    pairs = []
    for i in range(len(ll)- 1):
        for j in range (i+1, len(ll)):
            if ll[i] + ll[j] == target:
                pairs.append((f"{ll[i], ll[j]}"))


    return pairs

print(func([5,5,4,35,6], 10))
```

Find all triplets whose sum equals a target.

```python
def find_all_triplates(ll:list, target):
    pairs = []
    for i in range(len(ll)):
        for j in range(i+1, len(ll)):
            for k in range(j+1, len(ll)):
                if ll[i] + ll[j] + ll[k] == target:
                    pairs.append(f"{ll[i], ll[j], ll[k]}")


    return pairs


print(find_all_triplates([1,3,9,5,6,7,8], 18))
```

Flatten a nested list.

```python
def Flatten_list(ll:list):
    newlist = []
    for item in ll:
        if isinstance(item, list):
            newlist.extend(Flatten_list(item))
        else:
            newlist.append(item)

    return newlist

print(Flatten_list([2,3,6,[4,5,[25,12]]]))
```

Find the maximum value in a nested list.

```python
def find_max(ll:list):
    maximum = float('-inf')

    for item in ll:
        if isinstance(item, list):
            maximum = max(maximum, find_max(item))
        else:
            maximum = max(maximum,item)

    return maximum


print(find_max([2,3,6,[4,5,[25,12]]]))
```
