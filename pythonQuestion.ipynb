Count vowels in a string.

```python
vowels = ['a', 'e' ,'i','o','u', 'A','E','I','O','U']

def check_vowels(word:str):
    count = 0
    for c in word:
        if c in vowels:
            count += 1
        
    
    return count

print(check_vowels("Rahul"))
```

Count vowels and consonants separately.

```python
def vowels_consonent_count(word:str):
    vowels = 0
    consonent = 0

    for c in word:
        if c.isalpha():  
            if c in "aeiouAEIOU":
                vowels += 1
            else:
                consonent += 1
            
    return vowels, consonent


print(vowels_consonent_count("Rahul is a good boy"))
```

Find the frequency of each character.

```python
def frequency_of_charcter(string:str):
    data = {}
    for ch in string:
        if ch.isalpha():
            
            if ch not in data:
                data[ch] = 1
            else:
                data[ch] += 1
        
    return data

print(frequency_of_charcter("Rahul is a good boy"))
```

Find the first non-repeating character.

```python
def first_non_repeatingNumber(string:str):
    data = {}
    for ch in string:
        if ch.isalpha():
            if ch not in data:
                data[ch] = 1
            else:
                data[ch] += 1

    for key in data:
        if data[key] == 1:
            return key

print(first_non_repeatingNumber("Rahul is a good boy"))
```

Remove duplicate characters.

```python
def remove_duplicate_charcter(string:str):
    new_str = ""
    for ch in string:
        if ch not in new_str:
            new_str += ch
    
    return new_str


print(remove_duplicate_charcter("Rahulkumar"))
```

Find the longest word in a sentence.

```python
import re

def longest_word(sentence:str):
    words = sentence.split()
    longest_word = ""
    for word in words:
        if len(word) > len(longest_word):
            longest_word = word
    
    return longest_word
    
    

print(longest_word("Rahul is a good boy and lived in ahmedabad"))
```

Find the shortest word in a sentence.

```python
def sortest_word(sentences:str):
    words = sentences.split()
    sortest_word = words[0]
    
    for word in words:
        if word.isalpha():
            if word == sortest_word:
                continue
            elif len(sortest_word) > len(word):
                sortest_word = word


    return sortest_word

print(sortest_word("Rahul is a good boy")) 
```

Reverse every word in a sentence.

```python
def count_of_words(sentences:str):
    words = sentences.split()
    new_sentences = ""
    for word in words:
        new_sentences += " " + word[::-1]
    return new_sentences

print(count_of_words("Rahul is a good boy")) 
```

Reverse the order of words in a sentence.

```python
def count_of_words(sentences:str):
    words = sentences.split()
    words = words[::-1]
    new_sentence = ""
    for word in words:
        new_sentence += " " + word 
    
    return new_sentence
print(count_of_words("Rahul is a good boy")) 
```

Find all duplicate characters.

```python
def find_duplicateChar(words:str):
    data = {}
    for ch in words:
        if ch.isalpha():
            if ch not in data:
                data[ch] = 1
            else:
                data[ch] += 1

    duplicatechar = []
    for rep in data:
        
        if data[rep] > 1:
            duplicatechar.append(rep)
    return duplicatechar

print(find_duplicateChar("Rahul is a good boy"))
```

Find the most frequent character.

```python
def find_most_fequent_char(sentence:str):
    data = {}

    for ch in sentence:
        if ch.isalpha():
            if ch not in data:
                data[ch] = 1
            else:
                data[ch] += 1
    
    frequectchar = next(iter(data))
    for fre in data:
        if data[fre] <= data[frequectchar]:
            continue

        else:
            frequectchar = fre



    return frequectchar
    

print(find_most_fequent_char("Rahul is a good boy"))
```

Check whether two strings are anagrams.

```python
def check_anagrams(string1:str, string2: str):
    new_string1 = list(string1)
    new_string2 = list(string2)

    new_string1.sort()
    new_string2.sort()



    if new_string1 == new_string2:
        return True
    else:
        return False

print(check_anagrams("listen", "silent"))
```
